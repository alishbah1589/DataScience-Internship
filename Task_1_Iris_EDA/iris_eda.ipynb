# Iris Dataset Analysis

## Objective
The objective of this project is to explore and visualize the Iris dataset using Python libraries such as pandas, matplotlib, and seaborn.

## Dataset Description
The Iris dataset contains flower measurements including:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

The dataset also includes species classification for each flower.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## using scatter plot 
It hepls to analyze relationship between 2 variables
(Petal length vs petal width)

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='petal_length',
    y='petal_width',
    hue= 'species',
    data = df
)
plt.title("petal length vs petal width")
plt.xlabel("petal length")
plt.ylabel("petal width")

plt.show()
```

## Scatter Plot Insights

- Petal length and petal width show a positive relationship.
- Different iris species form distinct clusters.
- Setosa species appears clearly separated from the others.
- Petal measurements may be useful for species classification.

## Histogram
shows distribution of numerical values, 
how values spread
frequency of values
common ranges

```python
plt.figure(figsize=(8,6))
sns.histplot( 
    df['sepal_length'],
    bins=10,
    kde= True
)
plt.title("Sepal length Distribution")
plt.xlabel("sepal lenght")
plt.ylabel("frequency")
plt.show()
```

## Histogram Insights

- Sepal length values are concentrated around the middle range.
- The distribution appears moderately spread.
- Very small and very large sepal lengths are less frequent.

## Box plots 
help detect outliers and spread

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x='species',
    y='sepal_width',
    data=df
)

plt.title("Sepal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Width")

plt.show()
```

## Box Plot Insights

- Sepal width distribution differs across species.
- Some species show wider variability than others.
- A few potential outliers are visible in the dataset.

## Correlation Analysis


```python
correlation = df.drop('species', axis=1).corr()
correlation
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)
plt.title("correlation heatmap")
plt.show()
```

## Correlation Insights

- Petal length and petal width have a strong positive correlation.
- Sepal measurements show weaker relationships compared to petal measurements.
- Petal-related features may be more useful for species classification.

```python
sns.pairplot(df, hue='species')
plt.show()
```

## Pairplot Insights

- Petal-related features separate species more clearly than sepal-related features.
- Setosa species forms a clearly distinct cluster.
- Some overlap exists between Versicolor and Virginica species.
- Petal measurements appear highly useful for classification tasks.
